## 1.4. División en conjuntos de entrenamiento, validación y test

Separamos las columnas del dataset en dos variables:

- **`X`**: matriz de entrada, contiene las 4 características medidas de cada flor (longitud y anchura del sépalo y del pétalo). Tiene forma (150 filas, 4 columnas).
- **`y`**: vector de salida, contiene la etiqueta de clase de cada muestra (`Iris-setosa`, `Iris-versicolor` o `Iris-virginica`). Tiene longitud `150`.

A continuación dividimos los datos en tres subconjuntos:
- **Entrenamiento (64%)**: para ajustar los pesos del modelo.
- **Validación (16%)**: para supervisar el aprendizaje y detectar sobreajuste.
- **Test (20%)**: para evaluar el rendimiento final sobre datos no vistos.

````python
X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values

# Primera división: separar el 20% como test
X_temp, X_test, y_temp, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

# Segunda división: del 80% restante, 20% para validación
X_train, X_val, y_train, y_val = train_test_split(X_temp, y_temp, test_size=0.2, random_state=0)

print(f"Entrenamiento : {len(X_train)} muestras")
print(f"Validación    : {len(X_val)} muestras")
print(f"Test          : {len(X_test)} muestras")
````

## 1.5. Normalización de los datos

Normalizamos las características para que tengan media 0 y desviación estándar 1.
La normalización ayuda al modelo a converger más rápido durante el entrenamiento,
ya que evita que características con valores muy grandes dominen sobre las demás. La fórmula general de normalización es:

$$x_{\text{normalizado}} = \frac{x - \mu_{\text{train}}}{\sigma_{\text{train}}}$$

donde $\mu_{\text{train}}$ es la media y $\sigma_{\text{train}}$ es la desviación estándar
calculadas **únicamente sobre el conjunto de entrenamiento**. Aplicada al conjunto de entrenamiento, la fórmula queda:

$$X_{\text{train_norm}} = \frac{X_{\text{train}} - \mu_{\text{train}}}{\sigma_{\text{train}}}$$

El motivo de calcular $\mu$ y $\sigma$ solo con los datos de entrenamiento es evitar
la **fuga de información** (*data leakage*): si usáramos también los datos de validación
o test para calcular estos estadísticos, el modelo estaría indirectamente "viendo"
información de datos que se supone que no conoce, lo que daría una evaluación
artificialmente optimista. Los mismos valores de $\mu_{\text{train}}$ y $\sigma_{\text{train}}$
se aplican después a validación y test para mantener la coherencia.

````python
mu = np.mean(X_train, axis=0)
sigma = np.std(X_train, axis=0)
````

````python
X_train_n = (X_train - mu) / sigma
X_val_n = (X_val - mu) / sigma
X_test_n = (X_test - mu) / sigma

print(f"Media por característica : {mu}")
print(f"Desv. estándar : {sigma}")
````

## 1.6 Codificación one-hot de las etiquetas

La columna `label` del dataset Iris contiene cadenas de texto (`'Iris-setosa'`, `'Iris-versicolor'`,
`'Iris-virginica'`). Como se indica en el módulo 1, las redes neuronales trabajan con algoritmos matemáticos por lo que el MLP no puede trabajar directamente con texto. Necesitamos convertirlas a un formato numérico. Una opción sería asignar un número a cada clase
(0, 1, 2), pero esto introduciría un orden artificial entre las clases que no existe en realidad: el modelo podría interpretar que `Iris-virginica` (2) es "mayor" o "más
cercana" a `Iris-versicolor` (1) que a `Iris-setosa` (0), lo cual no tiene ningún sentido biológico.

La solución es la **codificación one-hot**: cada clase se representa como un vector
binario de longitud igual al número de clases, donde solo la posición correspondiente
a esa clase vale 1 y el resto valen 0:

$$\text{Iris-setosa} = \begin{pmatrix}1\\0\\0\end{pmatrix} \quad
\text{Iris-versicolor} = \begin{pmatrix}0\\1\\0\end{pmatrix} \quad
\text{Iris-virginica} = \begin{pmatrix}0\\0\\1\end{pmatrix}$$

Esta representación es importante porque la capa de salida del MLP
también produce un vector de 3 valores (uno por clase), y durante el entrenamiento necesitamos comparar esa salida con la etiqueta real. Con la codificación one-hot, esa comparación es directa: el modelo aprende a que la neurona de salida correcta se acerque a 1 y las demás se acerquen a 0.

El codificador se ajusta **únicamente con las etiquetas de entrenamiento** y después se aplica a validación y test, por el mismo motivo que con la normalización: evitar cualquier fuga de información.

````python
encoder = OneHotEncoder(sparse_output=False)
encoder.fit(y_train.reshape(-1, 1))

y_train_oh = encoder.transform(y_train.reshape(-1, 1)) # oh = One-Hot
y_val_oh = encoder.transform(y_val.reshape(-1, 1))
y_test_oh = encoder.transform(y_test.reshape(-1, 1))

print(f"Forma de y_train_oh: {y_train_oh.shape}  -> ({len(y_train)} muestras, 3 clases)")
print(f"\nAsí se ven las primeras 5 etiquetas del dataset codificadas:\n{y_train_oh[:5]}")
````

## 1.7. Funciones de activación

En el tema 4 del Módulo 1 se introduce el concepto de aprendizaje supervisado y
se mencionan las funciones de activación como parte necesaria de las redes neuronales. Aquí las retomamos y las situamos en contexto.

Pensemos primero en una sola neurona. Recibe varias entradas, las combina linealmente
(las multiplica por sus pesos y las suma) y obtiene un valor $z$. El problema es que
una combinación lineal de entradas solo puede representar relaciones lineales.
La **función de activación** $\phi(z)$ se aplica sobre ese valor $z$ para introducir
no linealidad, permitiendo que la red aprenda patrones más complejos:

$$\text{salida} = \phi\left(\sum_i w_i x_i\right)$$

En un MLP esta idea se repite en cada capa: cada neurona calcula su combinación
lineal y aplica una función de activación. Gracias a esto, al apilar varias capas,
la red puede aproximar funciones arbitrariamente complejas. Sin funciones de activación,
no importaría cuántas capas tuviéramos: el resultado seguiría siendo equivalente
a una única transformación lineal.

Las funciones que usaremos en este notebook son las siguientes:

````python
def sigmoid(z):
    """Función logística: comprime la entrada al intervalo (0, 1)."""
    return 1 / (1 + np.exp(-z))

def relu(z):
    """ReLU: devuelve 0 para valores negativos, el propio valor para positivos."""
    return np.maximum(0, z)

def tanh(z):
    """Tangente hiperbólica: comprime la entrada al intervalo (-1, 1)."""
    return np.tanh(z)

def softmax(z):
    """Softmax: convierte un vector de logits en una distribución de probabilidad."""
    exp_z = np.exp(z - np.max(z, axis=1, keepdims=True))  # estabilidad numérica
    return exp_z / np.sum(exp_z, axis=1, keepdims=True)
````

## 1.8. Definición de la arquitectura y parámetros del MLP

Un **MLP** es una red neuronal formada por varias capas de
neuronas conectadas entre sí. La información entra por la primera capa, se transforma
progresivamente a través de las capas ocultas y produce una predicción en la capa de salida.

En nuestro caso, el dataset Iris nos da 4 características de entrada que no podemos
modificar, son los datos que nos proporciona la naturaleza. Lo que sí podemos
manipular son los **pesos** $W$: cada entrada se multiplica por un peso antes de
pasar a la siguiente capa. El objetivo del entrenamiento es precisamente encontrar
los valores de esos pesos que hagan que la salida del modelo se parezca lo máximo
posible a la etiqueta real $y$. Dicho de forma simple: el modelo aprende ajustando $W$ hasta que, dado un vector
de entrada $X$, la salida $\hat{y}$ coincida con la clase correcta.

Las funciones de activación `sigmoid` en las capas ocultas introducen la no linealidad
necesaria para que el modelo pueda separar las tres clases de flores, que no son
linealmente separables en su totalidad. En la capa de salida usamos `softmax` porque
queremos obtener una probabilidad para cada una de las 3 clases, y softmax garantiza
que esas tres probabilidades sumen exactamente 1.

Los pesos se inicializan de forma aleatoria en el intervalo $[-0.1, 0.1]$ para
romper la simetría: si todos los pesos empezaran con el mismo valor, todas las
neuronas aprenderían exactamente lo mismo y la red no tendría capacidad real de
representación.

La arquitectura que construimos es:
```
Entrada (4) → Capa oculta 1 (16 neuronas, sigmoid) → Capa oculta 2 (8 neuronas, sigmoid) → Salida (3, softmax)
```

Sin título.jpg

````python
# Dimensiones de cada capa
n_input   = 4                     # 4 características. Las 4 columnas de información
n_hidden1 = 16                    # neuronas en la primera capa oculta
n_hidden2 = 8                     # neuronas en la segunda capa oculta
n_output  = 3                     # 3 clases

print(f"Arquitectura: {n_input} -> {n_hidden1} -> {n_hidden2} -> {n_output}")
````

````python
np.random.seed(42)

# Matrices de pesos e inicialización aleatoria
W1 = np.random.uniform(-0.1, 0.1, size=(n_input, n_hidden1))  # Pesos aleatorios de entrada 4  y salida 16 (4, 16)
W2 = np.random.uniform(-0.1, 0.1, size=(n_hidden1, n_hidden2))  # (16, 8)
W3 = np.random.uniform(-0.1, 0.1, size=(n_hidden2, n_output))   # (8, 3)

print(f"\nForma de W1: {W1.shape}")
print(f"Forma de W2: {W2.shape}")
print(f"Forma de W3: {W3.shape}")
````

## 1.9. Entrenamiento: forward propagation y backpropagation

El entrenamiento de un MLP se basa en repetir el siguiente ciclo durante un número
de épocas determinado. Una **época** es una pasada completa por todos los datos
de entrenamiento.

1. **Forward pass**: la entrada $X$ se propaga hacia adelante capa por capa.
   En cada capa se calcula la combinación lineal con los pesos y se aplica la
   función de activación. El resultado final es la predicción $\hat{y}$ del modelo.

2. **Función de pérdida**: comparamos la predicción $\hat{y}$ con la etiqueta
   real $y$ mediante la **entropía cruzada categórica**. Esta función mide cuánto
   se equivoca el modelo: cuanto mayor es el error, mayor es la pérdida.
   El objetivo del entrenamiento es minimizar este valor.

3. **Backpropagation**: una vez calculada la pérdida, propagamos el error hacia
   atrás a través de la red aplicando la regla de la cadena. Esto nos permite
   saber en qué medida ha contribuido cada peso al error total, es decir,
   calculamos el **gradiente** de la pérdida respecto a cada peso.

4. **Actualización de pesos**: ajustamos cada peso en la dirección opuesta al
   gradiente, con un paso controlado por la **tasa de aprendizaje**
   (*learning rate*) $\eta$:

$$W \leftarrow W - \eta \cdot \frac{\partial J}{\partial W}$$

Este proceso se repite época tras época. Con cada iteración los pesos se van
ajustando y la pérdida debería ir disminuyendo progresivamente, señal de que
el modelo está aprendiendo a relacionar las entradas con las clases correctas. Pueden consultar más información sobre el tema en el apartado 4.6 del módulo 1 de la asignatura.

````python
learning_rate = 0.01
epochs        = 300
loss_history  = []

for epoch in range(epochs):

    # ── FORWARD PROPAGATION ──────────────────────────────────────────────────
    # Capa oculta 1
    z1 = np.dot(X_train_n, W1)     # combinación lineal: (n_muestras, 16)
    a1 = sigmoid(z1)               # activación: (n_muestras, 16)

    # Capa oculta 2
    z2 = np.dot(a1, W2)            # combinación lineal: (n_muestras, 8)
    a2 = sigmoid(z2)               # activación: (n_muestras, 8)

    # Capa de salida
    z3 = np.dot(a2, W3)            # combinación lineal: (n_muestras, 3)
    a3 = softmax(z3)               # probabilidades de clase: (n_muestras, 3)

    # ── FUNCIÓN DE PÉRDIDA (entropía cruzada categórica) ─────────────────────
    loss = -np.sum(y_train_oh * np.log(a3 + 1e-9))
    loss_history.append(loss)

    # ── BACKPROPAGATION ───────────────────────────────────────────────────────
    # Error en la capa de salida
    delta3 = a3 - y_train_oh                                    # (n_muestras, 3)

    # Error en la capa oculta 2
    delta2 = np.dot(delta3, W3.T) * a2 * (1 - a2)              # (n_muestras, 8)

    # Error en la capa oculta 1
    delta1 = np.dot(delta2, W2.T) * a1 * (1 - a1)              # (n_muestras, 16)

    # ── ACTUALIZACIÓN DE PESOS ────────────────────────────────────────────────
    W3 -= learning_rate * np.dot(a2.T, delta3)
    W2 -= learning_rate * np.dot(a1.T, delta2)
    W1 -= learning_rate * np.dot(X_train_n.T, delta1)
````

Al finalizar el entrenamiento representamos la evolución de la pérdida a lo largo
de las épocas. Una curva que desciende de forma suave indica que el modelo está
aprendiendo correctamente. También imprimimos el valor de la pérdida al inicio
y al final para verificar que el entrenamiento ha reducido el error.

````python
# Curva de pérdida
plt.figure(figsize=(8, 4))
plt.plot(loss_history, color='steelblue', linewidth=2)
plt.title('Evolución de la pérdida durante el entrenamiento')
plt.xlabel('Época')
plt.ylabel('Pérdida (entropía cruzada)')
plt.grid(True, alpha=0.4)
plt.tight_layout()
plt.show()

print(f"Pérdida inicial: {loss_history[0]:.4f}")
print(f"Pérdida final  : {loss_history[-1]:.4f}")
````

## 1.10. Evaluación del modelo

Una vez entrenado el modelo, necesitamos medir qué tan bien generaliza sobre
datos que no ha visto durante el entrenamiento. Para ello usamos dos conjuntos:

- **Validación**: permite detectar problemas durante el entrenamiento, como el
  sobreajuste. Si la pérdida en entrenamiento baja pero el accuracy en validación
  no mejora, es una señal de que el modelo está memorizando los datos en lugar
  de aprender patrones generales. En este notebook no hemos usado la validación
  para tomar decisiones (como parar el entrenamiento antes o ajustar hiperparámetros),
  pero en un proyecto real sería el conjunto que guiaría esas decisiones.

- **Test**: es la evaluación final e imparcial del modelo. Solo se usa una vez,
  al final, para obtener una estimación honesta del rendimiento sobre datos completamente nuevos.

La métrica que usamos es el **accuracy** (exactitud), que mide la proporción de
muestras clasificadas correctamente:

$$\text{accuracy}(y, \hat{y}) = \frac{1}{n}\sum_{i=1}^{n} \mathbf{1}[y_i = \hat{y}_i]$$

donde $n$ es el número de muestras, $y_i$ es la etiqueta real y $\hat{y}_i$ es
la clase predicha por el modelo.

````python
def predict(X, W1, W2, W3):
    """Forward pass completo: devuelve la clase predicha para cada muestra."""
    a1 = sigmoid(np.dot(X,  W1))
    a2 = sigmoid(np.dot(a1, W2))
    a3 = softmax(np.dot(a2, W3))
    return np.argmax(a3, axis=1)

def accuracy(y_true, y_pred):
    return np.mean(y_true == y_pred)
````

````python
y_val_idx  = np.argmax(y_val_oh,  axis=1)
y_test_idx = np.argmax(y_test_oh, axis=1)

acc_val  = accuracy(y_val_idx,  predict(X_val_n,  W1, W2, W3))
acc_test = accuracy(y_test_idx, predict(X_test_n, W1, W2, W3))

print(f"Accuracy en validación : {acc_val:.4f}")
print(f"Accuracy en test       : {acc_test:.4f}")
````

Además del accuracy, visualizamos la **matriz de confusión**, que muestra de forma detallada cuántas muestras de cada clase fueron clasificadas correctamente y cuántas fueron confundidas con otra clase. Cada fila representa la clase real y cada columna la clase predicha: los valores en la diagonal son aciertos, y los valores fuera de la diagonal son errores.

Para entender la matriz de confusión es útil conocer los cuatro conceptos básicos que la componen (definidos aquí para una clase concreta frente al resto):

- **TP (True Positive)**: el modelo predice que la muestra pertenece a una clase y tiene razón.
- **TN (True Negative)**: el modelo predice que la muestra *no* pertenece a una clase y tiene razón.
- **FP (False Positive)**: el modelo predice que la muestra pertenece a una clase pero se equivoca.
- **FN (False Negative)**: el modelo predice que la muestra *no* pertenece a una clase pero se equivoca.

En un problema multiclase como Iris, estos valores se calculan de forma independiente para cada clase considerando las demás como negativas. Los valores en la diagonal de la matriz corresponden a los **TP** de cada clase, mientras que los valores fuera de la diagonal representan los **FP** y **FN** según la fila y columna en que aparezcan. El código a continuación, se aplica en el conjunto `test` ya que este conjunto se considera una prueba independiente.

Tengo un cuerpo técnico formado por 4 ojeadores [que serían la capa de entrada del modelo y cada uno hace referencia a una variable o atributo del equipo contra el cual jugaremos: altura media, velocidad media, pases y número de centros]. Luego tengo 2 capas ocultas con 3 analistas cada una y, finalmente, una capa de salida (yo, el entrenador) que decide la estrategia del partido [defensa, contraataque o presión alta].

Las matrices $W_1$, $W_2$ y $W_3$ representan el nivel de confianza o importancia (los pesos) que hay en las conexiones entre estos diferentes niveles de analistas.Antes de cada partido, los ojeadores traen los datos brutos (ej. 170cm de media, 40km/h, etc.).

A medida que esta información pasa por cada capa, los datos se transforman: los analistas no me dan datos sueltos, sino que toman la información de la capa anterior y la multiplican por sus "pesos" de confianza, creando conclusiones tácticas más complejas (ej. "si son bajos y rápidos, son peligrosos a mitad de campo").

Al llegar a mí (la última capa), yo no elijo un solo análisis y descarto el resto, sino que combino y sumo las opiniones de todos, dándole mucha más importancia a lo que dicen los analistas en los que tengo una confianza (peso) alta, y restando valor a los que tienen un peso bajo. Con esa mezcla final, organizo el partido.

Una vez finalizado el encuentro, según el resultado (el error), evalúo el trabajo de mis analistas y modifico la confianza (el peso) que tengo en cada uno. Esto sería el **Backpropagation**.

A medida que va avansando la pretemporada, la confianza en mis analistas se irá ajustando partido a partido.

En otras palabras, los "pesos" son la fuerza o importancia que tiene la información que transmite cada neurona.

Y el "aprendizaje" del modelo consiste precisamente en ir iterando y ajustando todos esos pesos durante el entrenamiento hasta encontrar la combinación de confianzas perfecta para acertar siempre con la táctica.

## ¿Qué pasaría si todos los pesos se inicializaran a 0?

En el caso de que empezara la pretemporada dándoles exactamente la misma confianza a todos los analistas (por ejemplo, inicializando todos los pesos a cero), se produciría un error grave de simetría.

Al realizar la autocrítica después de cada partido [backpropagation], la confianza o peso de cada uno de los analistas cambiaría exactamente de la misma forma. Esto provocaría que el aprendizaje no fuera real: todos actuarían como clones dando siempre la misma información (ya que, al pasar por la función de activación, el resultado sería idéntico para todos).

Como consecuencia, los analistas nunca se especializarían en leer diferentes partes del juego y el modelo fracasaría.

La métrica se define como:

$$\text{PCA}(y, \hat{y}) = \frac{1}{K} \sum_{k=1}^{K} \frac{\text{TP}_k^2}{\text{TP}_k^2 + \beta \cdot \text{FN}_k^2 + \gamma \cdot \text{FP}_k^2 + \delta \cdot \text{FN}_k \cdot \text{FP}_k}$$

- $\text{FP}_k$ es el número de muestras de otras clases clasificadas incorrectamente como clase $k$,
- $\beta = 2.0$ penaliza los falsos negativos al cuadrado,
- $\gamma = 1.5$ penaliza los falsos positivos al cuadrado,
- $\delta = 0.5$ penaliza la interacción entre falsos negativos y falsos positivos.
